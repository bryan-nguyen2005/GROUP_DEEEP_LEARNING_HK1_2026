## Nội dung bài

Hiển thị hai đồ thị $\sin(x)$ và $\cos(x)$ trong khoảng $x \in [-\pi, \pi]$.

## Bài làm

### Import thư viện

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Sinh dữ liệu hàm số

```python
X = np.linspace(-np.pi, np.pi, 256, endpoint=True)
C = np.cos(X)
S = np.sin(X)
```

### Kiểm tra dữ liệu

```python
print(f'+ X: len={len(X)}, range=[{X.min():.2f}, {X.max():.2f}], 10 samples={X[:10]}')
print(f'+ cos(X): len={len(C)}, range=[{C.min():.2f}, {C.max():.2f}], 10 samples={C[:10]}')
print(f'+ sin(X): len={len(S)}, range=[{S.min():.2f}, {S.max():.2f}], 10 samples={S[:10]}')
```

### Hiển thị đồ thị

```python
fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(X, C, color='blue', linewidth=2.5, label=r'$\cos(x)$')
ax.plot(X, S, color='red', linewidth=2.5, label=r'$\sin(x)$')

ax.set_xlim(X.min(), X.max())
ax.set_xticks(
    [-np.pi, -np.pi / 2, 0, np.pi / 2, np.pi],
    [r'$-\pi$', r'$-\pi/2$', r'$0$', r'$\pi/2$', r'$\pi$'],
)
ax.set_ylim(-1.1, 1.1)
ax.set_yticks([-1, -0.5, 0, 0.5, 1])
ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_title(r'Đồ thị $\sin(x)$ và $\cos(x)$ trên $[-\pi, \pi]$')
ax.axhline(0, color='black', linewidth=0.8)
ax.axvline(0, color='black', linewidth=0.8)
ax.grid(alpha=0.25)
ax.legend()
plt.show()
```
