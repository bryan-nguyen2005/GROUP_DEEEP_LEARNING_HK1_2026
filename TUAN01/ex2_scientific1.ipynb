## Nội dung bài

Hiển thị ngẫu nhiên $N=50$ hình tròn:

- Tâm $c_i=(x_i,y_i)$ ngẫu nhiên với $x_i, y_i \in [0,1]$.
- Màu sắc $color_i$ ngẫu nhiên với $color_i \in [0,1]$.
- Diện tích $area_i=\pi r_i^2$, trong đó $r_i \in [0,1]$ được sinh ngẫu nhiên.

## Bài làm

### Import thư viện

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Sinh dữ liệu ngẫu nhiên

Dùng bộ sinh số ngẫu nhiên của NumPy để tạo tọa độ tâm, màu sắc và bán kính của 50 hình tròn.

```python
N = 50
rng = np.random.default_rng()

x = rng.random(N)
y = rng.random(N)
colors = rng.random(N)
r = rng.random(N)
area = np.pi * r**2
```

### Kiểm tra dữ liệu

```python
print(f'+ x: len={len(x)}, range=[{x.min():.2f}, {x.max():.2f}], 10 samples={x[:10]}')
print(f'+ y: len={len(y)}, range=[{y.min():.2f}, {y.max():.2f}], 10 samples={y[:10]}')
print(f'+ colors: len={len(colors)}, range=[{colors.min():.2f}, {colors.max():.2f}], 10 samples={colors[:10]}')
print(f'+ r: len={len(r)}, range=[{r.min():.2f}, {r.max():.2f}], 10 samples={r[:10]}')
print(f'+ area: len={len(area)}, range=[{area.min():.2f}, {area.max():.2f}], 10 samples={area[:10]}')
```

### Hiển thị đồ thị

Tham số `s` của `scatter` được đo bằng point vuông. Vì vậy, diện tích toán học được nhân với một hệ số hiển thị để các hình tròn dễ quan sát, nhưng tỉ lệ diện tích giữa chúng vẫn được giữ nguyên.

```python
display_scale = 250

fig, ax = plt.subplots(figsize=(8, 6))
scatter = ax.scatter(
    x, y,
    s=display_scale * area,
    c=colors,
    cmap='viridis',
    alpha=0.55,
    edgecolors='black',
    linewidths=0.5,
)
ax.set(
    xlim=(0, 1),
    ylim=(0, 1),
    xlabel='$x$',
    ylabel='$y$',
    title='50 hình tròn ngẫu nhiên',
)
ax.set_aspect('equal', adjustable='box')
ax.grid(alpha=0.2)
fig.colorbar(scatter, ax=ax, label='Màu sắc ngẫu nhiên')
plt.show()
```
